```python
# create a notebook to connect to postgres SQL database using psycopg2 and .env file
# and create a table and insert data into the table
# and query the table and display the data
# and drop the table
# and close the connection

import psycopg2
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
conn	 = psycopg2.connect(
    dbname=os.getenv('PGDATABASE'),
    user=os.getenv('PGUSER'),
    password=os.getenv('PGPASSWORD'),
    host=os.getenv('PGHOST')
)
```

```python
# Step 4: Create a cursor object
cur = conn.cursor()
```

```python
# Step 5: Create a new table
cur.execute("""
    CREATE TABLE IF NOT EXISTS test_table (
        id SERIAL PRIMARY KEY,
        name VARCHAR(100),
        age INT
    )
""")
```

```python
conn.autocommit
```

```python
conn.commit()
```

- Autocommit là một cấu hình trong cơ sở dữ liệu để tự động lưu các thay đổi vào cơ sở dữ liệu sau mỗi câu lệnh SQL. Khi autocommit được bật, mỗi câu lệnh SQL sẽ được thực thi và lưu trực tiếp vào cơ sở dữ liệu mà không cần phải bắt đầu một giao dịch riêng biệt. Điều này có nghĩa là mọi thay đổi sẽ được áp dụng ngay lập tức và không thể hoàn tác.

- Bắt đầu transaction là một hành động để bắt đầu một giao dịch trong cơ sở dữ liệu. Một giao dịch là một tập hợp các câu lệnh SQL được thực thi như một đơn vị duy nhất. Khi bắt đầu một giao dịch, các thay đổi được thực hiện trong giao dịch sẽ không được áp dụng ngay lập tức vào cơ sở dữ liệu. Thay vào đó, chúng sẽ được lưu trữ tạm thời và chỉ được áp dụng khi giao dịch được xác nhận. Dùng `conn.cursor()` để bắt đầu transaction mới

- `Rollback` là một hành động để hủy bỏ một giao dịch và hoàn tác tất cả các thay đổi đã được thực hiện trong giao dịch đó. Khi một giao dịch được rollback, cơ sở dữ liệu sẽ trở về trạng thái trước khi giao dịch được bắt đầu, và mọi thay đổi sẽ bị hủy bỏ.

```python

cur = conn.cursor()

# Insert some data into table
cur.execute("""
    INSERT INTO test_table (name, age) VALUES
    ('John Doe', 30),
    ('Jane Doe', 25)
""")


cur.execute("SELECT * FROM test_table")
rows = cur.fetchall()
for row in rows:
    print(row)
```

```python
conn.commit()
```

```python
cur = conn.cursor()

cur.execute("DROP TABLE IF EXISTS test_table")
```
