# Join and Merge


```python
import pandas as pd

# Create sample datasets
person_df = pd.DataFrame({
    'id': [1, 2, 3, 4, 5, 6],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'],
    'age': [25, 30, 35, 40, 45, 50],
    'city_id': [1, 2, 3, 4, 5, None]  # Note: Frank has no city_id
})

city_df = pd.DataFrame({
    'city_id': [1, 2, 3, 6, 7],  # Note: city_id 6 and 7 don't match any person
    'city_name': ['New York', 'London', 'Paris', 'Berlin', 'Tokyo'],
    'country': ['USA', 'UK', 'France', 'Germany', 'Japan']
})
```

```python
person_df.dtypes
```

```python
person_df['city_id'] = person_df['city_id'].astype('Int64')
```

```python
person_df.dtypes
```

```python
city_df.dtypes
```

```python
print("Person DataFrame:")
print(person_df)
print("\nCity DataFrame:")
print(city_df)
```

```python
person_df
```

```python
city_df
```

```python
# Inner join
inner_join = pd.merge(person_df, city_df, on='city_id', how='inner')
print("\nInner Join:")
inner_join
```

```python
# Left join
left_join = pd.merge(person_df, city_df, on='city_id', how='left')
print("\nLeft Join:")
left_join
```

```python
# Right join
right_join = pd.merge(person_df, city_df, on='city_id', how='right')
print("\nRight Join:")
print(right_join)
```

```python
# Outer join
outer_join = pd.merge(person_df, city_df, on='city_id', how='outer')
print("\nOuter Join:")
print(outer_join)
```

```python
# Join with different column names
person_df_2 = person_df.rename(columns={'city_id': 'location_id'})
print("\nPerson DataFrame with different column name:")
print(person_df_2)
join_different_columns = pd.merge(person_df_2, city_df, left_on='location_id', right_on='city_id', how='left')
print("\nJoin with different column names:")
print(join_different_columns)
```

**Note**: 
- Ngoài `merge`, bạn có thể dùng dùng `join`, `concat` hoặc `query` để có kết quả tương tự. Nhưng Merge vẫn là dễ dùng nhất.
- Để xem thêm về Join, các bạn có thể tham khảo clip về Join này trong SQL https://youtu.be/PPS15CeGpjA. Syntax có khác nhau nhưng về bản chất join thì tương tự.
